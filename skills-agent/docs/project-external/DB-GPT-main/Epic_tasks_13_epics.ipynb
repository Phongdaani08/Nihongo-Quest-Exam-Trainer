# Epic 10: Text-to-SQL & Data Security

## Story 10.1: Generate dialect-specific SQL
- [Coding] Construct prompts from question, dialect, catalog, and retrieved context
- [Coding] Build SQL, assumptions, confidence, and referenced tables/columns
- [Coding] Repair SQL from normalized database errors
- [Testing] Test PostgreSQL syntax, ambiguity, and invalid structured output

## Story 10.2: Enforce SQL safety
- [Design] Define global datasource, schema, table, column, and PII policies
- [Coding] Parse SQL into an AST with SQLGlot
- [Coding] Allow only SELECT/read operations and reject multiple statements
- [Coding] Enforce allowlists, PII policy, SELECT-star policy, and LIMIT
- [Coding] Normalize SQL and create fingerprints
- [Testing] Build an adversarial SQL security suite

## Story 10.3: Execute queries safely
- [Coding] Execute with SQLAlchemy and psycopg using read-only transactions
- [Coding] Apply statement timeout, row limit, and byte limit
- [Coding] Add EXPLAIN-based query-cost checks and cancellation
- [Coding] Redact sensitive results and normalize database errors
- [Security] Prevent direct access to business-database credentials
- [Testing] Test DDL/DML, timeout, cancellation, permissions, and large results
